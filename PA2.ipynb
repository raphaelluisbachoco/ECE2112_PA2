## PROBLEM A: REPRODUCIBLE NORMALIZATION PROBLEM

```python
import numpy as np
```

```python
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))
```

```python
Mean = np.mean(X)
```

```python
print(Mean)
```

```python
Std = np.std(X)
```

```python
print(Std)
```

```python
A = X - Mean
```

```python
A
```

```python
X_normalized = A / Std
```

```python
X_normalized
```

```python
X_normalized_Mean = np.mean(X_normalized)
```

```python
X_normalized_Mean
```

```python
X_normalized_Std = np.std(X_normalized)
```

```python
X_normalized_Std
```

#### Required checks:






```python
X # Displays X
```

```python
X_normalized # Displays X_normalized
```

```python
X_normalized_Mean # Displays X normalized_mean
```

```python
X_normalized_Std # Displays X normalized_std
```

```python
np.save("X_normalized.npy",X_normalized)
```

## PROBLEM B: CUBES DIVISIBLE BY 4 PROBLEM

```python
A = np.arange(1,101,1)
A
```

```python
A ** 3
```

```python
C = A ** 3
```

```python
C = C.reshape(10,10)
```

```python
Z = C[C % 4 == 0]
Z
```

```python
div_by_4 = Z
```

#### Required Checks:

```python
C.shape # Displays C Shape
```

```python
div_by_4 # Display div_by_4 array
```

```python
div_by_4.size #Display div_by_4 Size
```

```python
np.save("div_by_4.npy",div_by_4)
```

## PROBLEM C: ABOVE-MEAN SQUARES PROBLEM

```python
B = np.arange(1,37,1)
S = (B ** 2).reshape(6,6)
S
```

```python
S.reshape(6, 6).shape
```

```python
S_mean = np.mean(S)
S_mean
```

```python
above_mean = S[S > S_mean]
```

#### Required Checks:

```python
S #Display S
```

```python
S_mean #Display S_mean
```

```python
above_mean #Display above_mean
```

```python
above_mean.size #Number of elements in above_mean
```

```python
np.save("above_mean.npy",above_mean)
```
